```python
import pandas as pd
```

```python
df = pd.read_csv('/retail_store_sales.csv')
```

```python
#Displaying the top 10 rows
print(df.head(10))
```

```python
print(df.shape)
```

```python
print(df.info())
```

```python
df["Item"].isnull().sum()
```

```python
df["Item"]=df["Item"].fillna("Unknown")
```

```python
df["Item"].isnull().sum()
```

```python
print(df.describe())
```

```python
df["Price Per Unit"].isnull().sum()
```

```python
df["Price Per Unit"]=df["Price Per Unit"].fillna(df["Price Per Unit"].median())
```

```python
df["Price Per Unit"].isnull().sum()
```

```python
df.head(10)
```

```python
df["Transaction Date"]=pd.to_datetime(df["Transaction Date"])
```

```python
df["Transaction Date"].dtype
```

```python
df.head(10)
```

```python
df["Discount Applied"].isnull().sum()
```

```python
df["Discount Applied"]=df["Discount Applied"].fillna("Unknown")
```

```python
df["Discount Applied"].isnull().sum()
```

```python
df.head(20)
```

```python
df["Quantity"]=df["Quantity"].fillna("NA")
df["Total Spent"]=df["Total Spent"].fillna("NA")
```

```python
df.head(10)
```

```python
df.to_csv('Retail_cleaned_dataset.csv',index=False)
```
